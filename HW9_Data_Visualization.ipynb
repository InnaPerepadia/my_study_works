Tips - це датасет, який містить інформацію про рахунки та чайові в ресторані.

**Колонки датасету:**
- `total_bill` - загальна сума рахунку (долари)

**Джерело:** Оригінальні дані зібрані в одному з американських ресторанів у 1990-х роках. Скачати і переглянути набір даних можна тут: https://www.kaggle.com/datasets/sakshisatre/tips-dataset

**1. Гістограма розподілу рахунків**
Завантажте датасет Tips (Чайові) з файла у матеріалах `tips.csv`.
Створіть гістограму для колонки `total_bill` з 20 bins.
Опціонально можна додати підписи осей ОХ, ОУ, а також назву графіку.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('C:/Users/XXX/Desktop/курс/1. Python/tips.csv')
```

```python
df['total_bill'].hist(bins=20, figsize=(10, 5), edgecolor="black")
plt.title('Розподіл рахунків в ресторані')
plt.xlabel('Сума рахунку')
plt.ylabel('Кількість рахунків');
```

Найбільша кількість рахунків на суму від 10 до 22 доларів.

---

**2. Box plot чайових за днями тижня**

Створіть box plot, який показує розподіл чайових (`tip`) за днями тижня (`day`).
В який день навище медіанне значення чайоивх? Зелена горизонтальна лінія всередині кожного box-plot.


```python
df.boxplot(column='tip', by='day', figsize=(10,5))
plt.title('Чайові за днями тижня')
plt.suptitle('')
plt.xlabel('День тижня')
plt.ylabel('Сума чайових')
plt.show()
```

Найвище медіанне значення чайових в неділю, найменше - в четвер.

---

**3. Scatter plot: рахунок vs чайові**

Створіть діаграму розсіювання для `total_bill` та `tip`.

```python
df.plot.scatter(x='tip', y='total_bill', alpha=0.5, figsize=(10,5))
plt.title('Залежність розміру чайових та суми рахунку')
plt.xlabel('Сума чайових')
plt.ylabel('Сума рахунку');
```

Чим вище розмір загального рахунку, ти більше розмір чайових.

```python
df.plot.scatter(x='tip', y='total_bill', alpha=0.5, c='sex',figsize=(10,5))
plt.title('Залежність розміру чайових та суми рахунку')
plt.xlabel('Сума чайових')
plt.ylabel('Сума рахунку');
```

```python
Найбільші чайові частіше давали чоловіки, також рахунки на більші суми частіше оплачували чоловіки.
```

---

**4. Bar chart: середній рахунок за часом**

Створіть стовпчасту діаграму середнього рахунку за часом дня (`time`).

```python
df.groupby('time')['total_bill'].mean().plot(kind='bar')
plt.title('Середній розмір рахунку за часом дня')
plt.xlabel('Час дня')
plt.ylabel('Сума рахунку');
```

**5. Інтерактивна діаграма розсіювання**

Створіть з допомогою бібліотек Plotly інтерактивний scatter plot для `total_bill` vs `tip` з кольором за статтю (`sex`).


```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
fig = px.scatter(df, x='total_bill', y='tip', color='sex',
                 title='Сума рахунку vs Сума чайових',
                 labels={'total_bill': 'Сума рахунку', 'tip': 'Сума чайових', 'sex': 'Стать'})
fig.show()
```

```python
#Графік з урахуванням розміру компанії

fig = px.scatter(df, x='tip', y='total_bill', color='sex', size='size',
                 title='Сума рахунку vs Сума чайових',
                 labels={'total_bill': 'Сума рахунку', 'tip': 'Сума чайових', 'sex': 'Стать'},
                 width=1000,
                 height=600)
fig.show()
```

**6. Інтерактивний box plot: чайові за днями та статтю**

Створіть інтерактивний box plot для чайових за днями тижня з розбивкою за кольором по статі.

```python
fig_2 = px.box(df, x='day', y='tip', color='sex',
               title='Розподіл чайових по дням та за статтю',
               labels={'day':'День тижня', 'tip':'Сума чайових', 'sex':'Стать'})
fig_2.show()
```
